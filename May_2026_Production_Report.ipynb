```python
# 5. DATA QUALITY VALIDATION

# 1. Validate dataset dimensions
print("=" * 60)
print("DATA QUALITY VALIDATION REPORT")
print("=" * 60)

print(f"\nTotal records       : {len(df)}")
print(f"Total columns       : {len(df.columns)}")
print(f"Duplicate rows      : {df.duplicated().sum()}")

# 2. Validate column names
expected_columns = [
    "Report Date", "LINE", "HMR", "LCR", "PCR",
    "WASTAGE", "% M/C", "PROD", "PKG"
]

print("\nColumn structure:")
print("Expected columns match:", list(df.columns) == expected_columns)

# 3. Check missing values
missing_report = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing Percentage": (df.isna().mean() * 100).round(2)
})

print("\nMissing Values Report:")
display(missing_report)

# 4. Check data types
print("\nData Types:")
display(df.dtypes.rename("Data Type").to_frame())

# 5. Check report date coverage
dates = pd.to_datetime(df["Report Date"], errors="coerce")

print("\nDate Validation:")
print("Invalid or missing dates:", dates.isna().sum())
print("Unique report dates:", dates.nunique())

if dates.notna().any():
    print("First report date:", dates.min().date())
    print("Last report date :", dates.max().date())

# 6. Check line coverage
print("\nLine Validation:")
print("Unique lines:", sorted(df["LINE"].dropna().unique().tolist()))
print("Missing line values:", df["LINE"].isna().sum())

print("\nValidation report completed.")
```

```python
# 7. VALIDATE MISSING PROD AND PKG RECORDS

missing_rows = df.loc[
    df["PROD"].isna() | df["PKG"].isna(),
    ["Report Date", "LINE", "PROD", "PKG"]
].copy()

print("Rows with missing PROD or PKG:", len(missing_rows))

print("\nLine-wise missing value summary:")
display(
    df.groupby("LINE")[["PROD", "PKG"]]
      .apply(lambda group: group.isna().sum())
      .rename(columns={
          "PROD": "Missing PROD",
          "PKG": "Missing PKG"
      })
)

print("\nMissing records preview:")
display(missing_rows.head(10))
```

```python
# 8. PRODUCTION CALCULATION VALIDATION

# Calculate expected machine wastage
expected_wastage = df["LCR"] - df["PCR"]

# Identify records where the reported wastage differs
wastage_mismatch = df.loc[
    df["WASTAGE"] != expected_wastage,
    ["Report Date", "LINE", "LCR", "PCR", "WASTAGE"]
].copy()

print("Wastage calculation validation")
print("-" * 45)
print("Total records:", len(df))
print("Mismatched records:", len(wastage_mismatch))

if wastage_mismatch.empty:
    print("Status: PASSED — WASTAGE equals LCR - PCR.")
else:
    print("Status: REVIEW REQUIRED")
    display(wastage_mismatch.head(20))


# Calculate expected machine wastage
expected_wastage = df["LCR"] - df["PCR"]

# Identify records where the reported wastage differs
wastage_mismatch = df.loc[
    df["WASTAGE"] != expected_wastage,
    ["Report Date", "LINE", "LCR", "PCR", "WASTAGE"]
].copy()

print("Wastage calculation validation")
print("-" * 45)
print("Total records:", len(df))
print("Mismatched records:", len(wastage_mismatch))

if wastage_mismatch.empty:
    print("Status: PASSED — WASTAGE equals LCR - PCR.")
else:
    print("Status: REVIEW REQUIRED")
    display(wastage_mismatch.head(20))


# Calculate expected machine wastage
expected_wastage = df["LCR"] - df["PCR"]

# Identify records where the reported wastage differs
wastage_mismatch = df.loc[
    df["WASTAGE"] != expected_wastage,
    ["Report Date", "LINE", "LCR", "PCR", "WASTAGE"]
].copy()

print("Wastage calculation validation")
print("-" * 45)
print("Total records:", len(df))
print("Mismatched records:", len(wastage_mismatch))

if wastage_mismatch.empty:
    print("Status: PASSED — WASTAGE equals LCR - PCR.")
else:
    print("Status: REVIEW REQUIRED")
    display(wastage_mismatch.head(20))
```

```python
# 9. MACHINE WASTAGE PERCENTAGE VALIDATION

expected_percentage = (df["WASTAGE"] / df["LCR"]) * 100

# Allow a small tolerance for decimal rounding
percentage_mismatch = df.loc[
    ~np.isclose(
        df["% M/C"],
        expected_percentage,
        rtol=1e-05,
        atol=0.01,
        equal_nan=False
    ),
    ["Report Date", "LINE", "LCR", "WASTAGE", "% M/C"]
].copy()

print("Machine wastage percentage validation")
print("-" * 45)
print("Total records:", len(df))
print("Mismatched records:", len(percentage_mismatch))

if percentage_mismatch.empty:
    print("Status: PASSED")
else:
    print("Status: REVIEW REQUIRED")
    display(percentage_mismatch.head(20))
```

```python
# 10. OPERATIONAL WASTAGE VALIDATION

valid_pkg = df["PKG"].notna()

operational_wastage = (
    df.loc[valid_pkg, "PCR"] -
    df.loc[valid_pkg, "PKG"]
)

# Check that operational wastage is non-negative
negative_operational_wastage = df.loc[
    valid_pkg & ((df["PCR"] - df["PKG"]) < 0),
    ["Report Date", "LINE", "PCR", "PKG"]
].copy()

print("Operational wastage validation")
print("-" * 45)
print("Records with PKG available:", valid_pkg.sum())
print("Records with PKG missing:", (~valid_pkg).sum())
print(
    "Negative operational wastage records:",
    len(negative_operational_wastage)
)

if negative_operational_wastage.empty:
    print("Status: PASSED — no negative operational wastage found.")
else:
    print("Status: REVIEW REQUIRED")
    display(negative_operational_wastage.head(20))
```

```python
# 11. FINAL DATA QUALITY SUMMARY

total_records = len(df)
total_columns = len(df.columns)
duplicate_rows = df.duplicated().sum()
missing_cells = df.isna().sum().sum()
invalid_dates = pd.to_datetime(
    df["Report Date"], errors="coerce"
).isna().sum()

summary = pd.DataFrame({
    "Validation Check": [
        "Total records",
        "Total columns",
        "Duplicate rows",
        "Missing cells",
        "Invalid report dates",
        "Unique report dates",
        "Unique production lines",
        "Wastage formula mismatches",
        "Machine wastage percentage mismatches",
        "Negative operational wastage records"
    ],
    "Result": [
        total_records,
        total_columns,
        duplicate_rows,
        missing_cells,
        invalid_dates,
        df["Report Date"].nunique(),
        df["LINE"].nunique(),
        len(wastage_mismatch),
        len(percentage_mismatch),
        len(negative_operational_wastage)
    ]
})

display(summary)
```

```python
# 14. INITIALIZE PRODUCTION ANALYSIS

import pandas as pd
import numpy as np

file_path = "May_2026_Validated_Master_Dataset.xlsx"

production_df = pd.read_excel(
    file_path,
    sheet_name="Master Dataset"
)

# Ensure report dates use datetime format
production_df["Report Date"] = pd.to_datetime(
    production_df["Report Date"],
    errors="coerce"
)

print("Production analysis dataset loaded.")
print("-" * 45)
print("Records:", len(production_df))
print("Columns:", len(production_df.columns))
print("Date range:",
      production_df["Report Date"].min().date(),
      "to",
      production_df["Report Date"].max().date())
print("Production lines:",
      sorted(production_df["LINE"].dropna().unique().tolist()))

display(production_df.head())
```

```python
# 15. RECORD COUNT VALIDATION BY PRODUCTION LINE

line_validation = (
    production_df.groupby("LINE")
    .agg(
        Total_Records=("Report Date", "size"),
        Unique_Dates=("Report Date", "nunique")
    )
    .reset_index()
)

line_validation["Expected_Records"] = 30
line_validation["Validation_Status"] = np.where(
    (line_validation["Total_Records"] == 30) &
    (line_validation["Unique_Dates"] == 30),
    "PASS",
    "REVIEW"
)

display(line_validation)

print("\nOverall validation:")
print("Total records:", production_df.shape[0])
print("Lines checked:", production_df["LINE"].nunique())
print(
    "Lines passing validation:",
    (line_validation["Validation_Status"] == "PASS").sum()
)
```

```python
# 16. DAILY PRODUCTION LINE VALIDATION

daily_line_check = (
    production_df.groupby("Report Date")
    .agg(
        Total_Records=("LINE", "size"),
        Unique_Lines=("LINE", "nunique")
    )
    .reset_index()
)

daily_line_check["Status"] = np.where(
    (daily_line_check["Total_Records"] == 9) &
    (daily_line_check["Unique_Lines"] == 9),
    "PASS",
    "REVIEW"
)

display(daily_line_check)

print("\nDaily validation summary")
print("-" * 35)
print("Dates checked:", len(daily_line_check))
print("Dates passed:", (daily_line_check["Status"] == "PASS").sum())
print("Dates requiring review:", (daily_line_check["Status"] == "REVIEW").sum())
```

```python
# 17. DUPLICATE DATE-LINE VALIDATION

duplicate_date_line = production_df[
    production_df.duplicated(
        subset=["Report Date", "LINE"],
        keep=False
    )
].sort_values(["Report Date", "LINE"])

print("Duplicate date-line combinations:",
      len(duplicate_date_line))

if duplicate_date_line.empty:
    print("Status: PASSED — each date-line combination is unique.")
else:
    print("Status: REVIEW REQUIRED")
    display(
        duplicate_date_line[
            ["Report Date", "LINE", "HMR", "LCR", "PCR"]
        ]
    )
```

```python
# 18. PRODUCTION QUANTITY VALIDATION

quantity_columns = ["HMR", "LCR", "PCR", "WASTAGE", "PROD", "PKG"]

quantity_validation = []

for column in quantity_columns:
    values = pd.to_numeric(production_df[column], errors="coerce")

    quantity_validation.append({
        "Column": column,
        "Negative_Values": int((values < 0).sum()),
        "Missing_Values": int(values.isna().sum()),
        "Status": (
            "PASS"
            if not (values < 0).any()
            else "REVIEW"
        )
    })

quantity_report = pd.DataFrame(quantity_validation)

display(quantity_report)
```

```python
# 19. TOTAL WASTAGE VALIDATION

valid_rows = production_df["PKG"].notna()

expected_total_wastage = (
    production_df.loc[valid_rows, "WASTAGE"]
    + production_df.loc[valid_rows, "PCR"]
    - production_df.loc[valid_rows, "PKG"]
)

calculated_total_wastage = (
    production_df.loc[valid_rows, "LCR"]
    - production_df.loc[valid_rows, "PKG"]
)

total_wastage_mismatch = (
    expected_total_wastage != calculated_total_wastage
)

print("Total wastage validation")
print("-" * 40)
print("Records checked:", int(valid_rows.sum()))
print("Mismatches:", int(total_wastage_mismatch.sum()))

if not total_wastage_mismatch.any():
    print("Status: PASSED")
else:
    print("Status: REVIEW REQUIRED")
```

```python
# 21. CONSOLIDATED VALIDATION STATUS

validation_summary = pd.DataFrame([
    ["Dataset structure", "270 records, 9 columns", "PASS"],
    ["Date coverage", "30 report dates", "PASS"],
    ["Line coverage", "9 lines per date", "PASS"],
    ["Duplicate full rows", "0 duplicates", "PASS"],
    ["Duplicate date-line pairs", "0 duplicates", "PASS"],
    ["Invalid report dates", "0", "PASS"],
    ["Negative quantities", "0", "PASS"],
    ["Machine wastage formula", "LCR - PCR", "PASS"],
    ["Machine wastage percentage", "Formula verified", "PASS"],
    ["Operational wastage", "No negative values", "PASS"],
    ["Total wastage formula", "LCR - PKG", "PASS"],
    ["Packing vs production", "No PKG > PROD records", "PASS"],
    ["Line 9 PROD/PKG values", "Missing; source review required", "REVIEW"]
], columns=["Validation Check", "Result", "Status"])

display(validation_summary)
```

```python
# 22. DAILY PRODUCTION SUMMARY

daily_summary = (
    production_df.groupby("Report Date")
    .agg(
        Total_HMR=("HMR", "sum"),
        Total_LCR=("LCR", "sum"),
        Total_PCR=("PCR", "sum"),
        Total_Machine_Wastage=("WASTAGE", "sum"),
        Total_PROD=("PROD", "sum"),
        Total_PKG=("PKG", "sum"),
        Lines_Reported=("LINE", "nunique")
    )
    .reset_index()
)

# Calculate total wastage where the required values are available
daily_summary["Total_Wastage"] = (
    daily_summary["Total_LCR"] - daily_summary["Total_PKG"]
)

# Calculate machine wastage percentage
daily_summary["Machine_Wastage_Percentage"] = np.where(
    daily_summary["Total_LCR"] != 0,
    daily_summary["Total_Machine_Wastage"]
    / daily_summary["Total_LCR"] * 100,
    np.nan
)

display(daily_summary.head(10))

print("Daily summary created.")
print("Number of daily records:", len(daily_summary))
```

```python
# 23. MONTHLY PRODUCTION SUMMARY BY LINE

line_summary = (
    production_df.groupby("LINE")
    .agg(
        Days_Reported=("Report Date", "nunique"),
        Total_HMR=("HMR", "sum"),
        Total_LCR=("LCR", "sum"),
        Total_PCR=("PCR", "sum"),
        Total_Machine_Wastage=("WASTAGE", "sum"),
        Total_PROD=("PROD", "sum"),
        Total_PKG=("PKG", "sum")
    )
    .reset_index()
)

# Calculate machine wastage percentage
line_summary["Machine_Wastage_Percentage"] = np.where(
    line_summary["Total_LCR"] != 0,
    line_summary["Total_Machine_Wastage"]
    / line_summary["Total_LCR"] * 100,
    np.nan
)

# Identify completeness of production and packing data
line_summary["PROD_Missing_Days"] = (
    production_df.groupby("LINE")["PROD"]
    .apply(lambda s: s.isna().sum())
    .to_numpy()
)

line_summary["PKG_Missing_Days"] = (
    production_df.groupby("LINE")["PKG"]
    .apply(lambda s: s.isna().sum())
    .to_numpy()
)

display(line_summary.round(2))

print("\nMonthly line summary created.")
print("Lines included:", len(line_summary))
```

```python
# 24. CORRECT MISSING TOTALS IN LINE SUMMARY

for line in [9]:
    missing_prod = line_summary.loc[
        line_summary["LINE"] == line, "PROD_Missing_Days"
    ].iloc[0]

    missing_pkg = line_summary.loc[
        line_summary["LINE"] == line, "PKG_Missing_Days"
    ].iloc[0]

    if missing_prod > 0:
        line_summary.loc[
            line_summary["LINE"] == line, "Total_PROD"
        ] = np.nan

    if missing_pkg > 0:
        line_summary.loc[
            line_summary["LINE"] == line, "Total_PKG"
        ] = np.nan

display(
    line_summary[
        [
            "LINE",
            "Days_Reported",
            "Total_LCR",
            "Total_PCR",
            "Total_Machine_Wastage",
            "Total_PROD",
            "Total_PKG",
            "PROD_Missing_Days",
            "PKG_Missing_Days"
        ]
    ].round(2)
)
```

```python
# 25. LINE-WISE DATA COMPLETENESS REPORT


completeness_report = (
    production_df.groupby("LINE")
    .agg(
        Total_Records=("Report Date", "size"),
        PROD_Available=("PROD", "count"),
        PKG_Available=("PKG", "count")
    )
    .reset_index()
)

completeness_report["PROD_Completeness_%"] = (
    completeness_report["PROD_Available"]
    / completeness_report["Total_Records"] * 100
).round(2)

completeness_report["PKG_Completeness_%"] = (
    completeness_report["PKG_Available"]
    / completeness_report["Total_Records"] * 100
).round(2)

display(completeness_report)

print("\nData completeness report created.")
```

```python
# 26. DAILY DATA COMPLETENESS STATUS

daily_completeness = (
    production_df.groupby("Report Date")
    .agg(
        Total_Lines=("LINE", "nunique"),
        PROD_Missing=("PROD", lambda s: s.isna().sum()),
        PKG_Missing=("PKG", lambda s: s.isna().sum())
    )
    .reset_index()
)

daily_completeness["PROD_Status"] = np.where(
    daily_completeness["PROD_Missing"] == 0,
    "Complete",
    "Incomplete"
)

daily_completeness["PKG_Status"] = np.where(
    daily_completeness["PKG_Missing"] == 0,
    "Complete",
    "Incomplete"
)

display(daily_completeness.head(10))

print("\nDaily completeness summary")
print("Dates checked:", len(daily_completeness))
print(
    "Dates with incomplete PROD:",
    (daily_completeness["PROD_Status"] == "Incomplete").sum()
)
print(
    "Dates with incomplete PKG:",
    (daily_completeness["PKG_Status"] == "Incomplete").sum()
)
```

```python
# 27. EXPORT VALIDATION AND SUMMARY REPORTS
from pathlib import Path

report_file = Path("May_2026_Production_Summary_Reports.xlsx")

with pd.ExcelWriter(report_file, engine="openpyxl") as writer:
    validation_summary.to_excel(
        writer, sheet_name="Validation Summary", index=False
    )
    line_summary.to_excel(
        writer, sheet_name="Monthly Line Summary", index=False
    )
    completeness_report.to_excel(
        writer, sheet_name="Line Completeness", index=False
    )
    daily_completeness.to_excel(
        writer, sheet_name="Daily Completeness", index=False
    )
    daily_summary.to_excel(
        writer, sheet_name="Daily Summary", index=False
    )

print("Summary reports exported successfully.")
print("File:", report_file.resolve())
print("Sheets: 5")
```

```python
# 29. DAILY PRODUCTION PERFORMANCE

daily_performance = (
    production_df.groupby("Report Date")
    .agg(
        Total_LCR=("LCR", "sum"),
        Total_PCR=("PCR", "sum"),
        Total_Wastage=("WASTAGE", "sum"),
        Available_PROD=("PROD", "sum"),
        Available_PKG=("PKG", "sum")
    )
    .reset_index()
)

# Calculate machine wastage percentage
daily_performance["Machine_Wastage_%"] = np.where(
    daily_performance["Total_LCR"] != 0,
    daily_performance["Total_Wastage"]
    / daily_performance["Total_LCR"] * 100,
    np.nan
)

# Mark production and packing totals as incomplete when values are missing
daily_performance["PROD_Complete"] = (
    production_df.groupby("Report Date")["PROD"]
    .apply(lambda s: s.notna().all())
    .to_numpy()
)

daily_performance["PKG_Complete"] = (
    production_df.groupby("Report Date")["PKG"]
    .apply(lambda s: s.notna().all())
    .to_numpy()
)

daily_performance.loc[
    ~daily_performance["PROD_Complete"], "Available_PROD"
] = np.nan

daily_performance.loc[
    ~daily_performance["PKG_Complete"], "Available_PKG"
] = np.nan

display(daily_performance.head(10))

print("\nDaily performance table created.")
print("Days analyzed:", len(daily_performance))
```

```python
# 30. MONTHLY LINE-WISE PERFORMANCE

line_performance = (
    production_df.groupby("LINE")
    .agg(
        Days_Reported=("Report Date", "nunique"),
        Total_LCR=("LCR", "sum"),
        Total_PCR=("PCR", "sum"),
        Total_Machine_Wastage=("WASTAGE", "sum"),
        Available_PROD=("PROD", "sum"),
        Available_PKG=("PKG", "sum"),
        Missing_PROD=("PROD", lambda s: s.isna().sum()),
        Missing_PKG=("PKG", lambda s: s.isna().sum())
    )
    .reset_index()
)

# Calculate machine wastage percentage
line_performance["Machine_Wastage_%"] = np.where(
    line_performance["Total_LCR"] != 0,
    line_performance["Total_Machine_Wastage"]
    / line_performance["Total_LCR"] * 100,
    np.nan
)

# Do not present incomplete totals as complete totals
line_performance.loc[
    line_performance["Missing_PROD"] > 0, "Available_PROD"
] = np.nan

line_performance.loc[
    line_performance["Missing_PKG"] > 0, "Available_PKG"
] = np.nan

display(line_performance.round(2))

print("\nMonthly line-wise performance created.")
```

```python
# 31. HIGHLIGHT HIGHEST MACHINE WASTAGE

from IPython.display import display

# Find the maximum machine wastage
max_wastage = line_performance["Total_Machine_Wastage"].max()

# Highlight every row tied for the highest value
def highlight_max_wastage(row):
    if row["Total_Machine_Wastage"] == max_wastage:
        return [
            "background-color: yellow; color: black; font-weight: bold"
        ] * len(row)
    return [""] * len(row)

styled_line_performance = (
    line_performance.style
    .format(precision=2)
    .apply(highlight_max_wastage, axis=1)
)

display(styled_line_performance)

print("Highest total machine wastage:", max_wastage)
print(
    "Line(s) with highest machine wastage:",
    line_performance.loc[
        line_performance["Total_Machine_Wastage"] == max_wastage,
        "LINE"
    ].tolist()
)
```

```python
# 34. OPERATIONAL WASTAGE AND PERCENTAGE BY LINE

operational_df = production_df.copy()

# Calculate total wastage
operational_df["Total_Wastage"] = (
    operational_df["LCR"] - operational_df["PKG"]
)

# Calculate operational wastage
operational_df["Operational_Wastage"] = (
    operational_df["Total_Wastage"]
    - operational_df["WASTAGE"]
)

# Monthly summary by production line
operational_summary = (
    operational_df.groupby("LINE")
    .agg(
        Days_Reported=("Report Date", "nunique"),
        Total_PCR=("PCR", "sum"),
        Total_PKG=("PKG", "sum"),
        Total_Wastage=("Total_Wastage", "sum"),
        Machine_Wastage=("WASTAGE", "sum"),
        Operational_Wastage=("Operational_Wastage", "sum"),
        Missing_PKG=("PKG", lambda s: s.isna().sum())
    )
    .reset_index()
)

# Preserve unknown totals for lines with missing PKG
missing_pkg = operational_summary["Missing_PKG"] > 0

operational_summary.loc[
    missing_pkg,
    ["Total_PKG", "Total_Wastage", "Operational_Wastage"]
] = np.nan

# Operational wastage percentage relative to PCR
operational_summary["Operational_Wastage_%"] = np.where(
    operational_summary["Total_PCR"] > 0,
    operational_summary["Operational_Wastage"]
    / operational_summary["Total_PCR"] * 100,
    np.nan
)

# Identify highest values, excluding missing results
max_operational = operational_summary["Operational_Wastage"].max()
max_operational_pct = operational_summary["Operational_Wastage_%"].max()

# Highlight highest quantity in yellow and highest percentage in green
def highlight_operational(row):
    styles = [""] * len(row)

    if pd.notna(max_operational) and row["Operational_Wastage"] == max_operational:
        styles = [
            "background-color: yellow; color: black; font-weight: bold"
        ] * len(row)

    if (
        pd.notna(max_operational_pct)
        and row["Operational_Wastage_%"] == max_operational_pct
    ):
        styles = [
            "background-color: lightgreen; color: black; font-weight: bold"
        ] * len(row)

    return styles

display(
    operational_summary.style
    .format(precision=2, na_rep="N/A")
    .apply(highlight_operational, axis=1)
)

print("Highest operational wastage:", max_operational)
print(
    "Line(s):",
    operational_summary.loc[
        operational_summary["Operational_Wastage"] == max_operational,
        "LINE"
    ].tolist()
)

print("\nHighest operational wastage percentage:", round(max_operational_pct, 2), "%")
print(
    "Line(s):",
    operational_summary.loc[
        operational_summary["Operational_Wastage_%"] == max_operational_pct,
        "LINE"
    ].tolist()
)
```

```python
# 38. HIGHLIGHT HIGHEST AND LOWEST EFFICIENCY BY LINE


from IPython.display import display

metric_columns = [
    "PKG_vs_PROD_%",
    "PCR_vs_LCR_%",
    "PROD_per_HMR",
    "PKG_vs_LCR_%"
]

def highlight_extremes(column):
    styles = pd.Series("", index=column.index)

    valid = column.dropna()

    if valid.empty:
        return styles

    # Highest value = green
    styles.loc[valid.index[valid == valid.max()]] = (
        "background-color: lightgreen; color: black; font-weight: bold"
    )

    # Lowest value = yellow
    styles.loc[valid.index[valid == valid.min()]] = (
        "background-color: yellow; color: black; font-weight: bold"
    )

    return styles

styled_efficiency = (
    efficiency_report.style
    .format(precision=2, na_rep="N/A")
    .apply(
        lambda col: highlight_extremes(col)
        if col.name in metric_columns
        else pd.Series("", index=col.index),
        axis=0
    )
)

display(styled_efficiency)
```

```python
# 39. MONTHLY PRODUCTION PER HOUR BY LINE

production_per_hour = (
    production_df.groupby("LINE")
    .agg(
        Total_HMR=("HMR", "sum"),
        Total_PROD=("PROD", "sum"),
        Missing_PROD=("PROD", lambda s: s.isna().sum()),
        Days_Reported=("Report Date", "nunique")
    )
    .reset_index()
)

# Do not report a complete rate if production data is missing
production_per_hour.loc[
    production_per_hour["Missing_PROD"] > 0,
    "Total_PROD"
] = np.nan

production_per_hour["Production_Per_Hour"] = np.where(
    (production_per_hour["Total_HMR"] > 0)
    & production_per_hour["Total_PROD"].notna(),
    production_per_hour["Total_PROD"]
    / production_per_hour["Total_HMR"],
    np.nan
)

# Highlight the highest and lowest production rates
valid_rates = production_per_hour["Production_Per_Hour"].dropna()
highest_rate = valid_rates.max()
lowest_rate = valid_rates.min()

def highlight_production_rate(row):
    styles = [""] * len(row)
    rate = row["Production_Per_Hour"]

    if pd.notna(rate) and rate == highest_rate:
        styles = [
            "background-color: lightgreen; color: black; font-weight: bold"
        ] * len(row)
    elif pd.notna(rate) and rate == lowest_rate:
        styles = [
            "background-color: yellow; color: black; font-weight: bold"
        ] * len(row)

    return styles

display(
    production_per_hour.style
    .format(precision=2, na_rep="N/A")
    .apply(highlight_production_rate, axis=1)
)

print("Highest production per hour:", round(highest_rate, 2))
print(
    "Line(s):",
    production_per_hour.loc[
        production_per_hour["Production_Per_Hour"] == highest_rate,
        "LINE"
    ].tolist()
)

print("\nLowest production per hour:", round(lowest_rate, 2))
print(
    "Line(s):",
    production_per_hour.loc[
        production_per_hour["Production_Per_Hour"] == lowest_rate,
        "LINE"
    ].tolist()
)
```

```python
# STEP 3: PERFORMANCE-BASED HIGHLIGHTING

from openpyxl import load_workbook
from openpyxl.styles import PatternFill, Font
from pathlib import Path
import math

file_path = Path("May_2026_Production_Metrics.xlsx")
wb = load_workbook(file_path)

green_fill = PatternFill("solid", fgColor="C6EFCE")
red_fill = PatternFill("solid", fgColor="FFC7CE")

wastage_terms = ["wastage"]
efficiency_terms = [
    "production_per_hour",
    "prod_per_hmr",
    "pkg_vs_prod",
    "pcr_vs_lcr",
    "pkg_vs_lcr"
]

for ws in wb.worksheets:
    if ws.max_row < 3:
        continue

    for col_idx in range(1, ws.max_column + 1):
        header = str(ws.cell(1, col_idx).value or "").lower().replace(" ", "_")

        is_wastage = any(term in header for term in wastage_terms)
        is_efficiency = any(term in header for term in efficiency_terms)

        if not (is_wastage or is_efficiency):
            continue

        values = []
        for row_idx in range(2, ws.max_row + 1):
            value = ws.cell(row_idx, col_idx).value
            if isinstance(value, (int, float)) and not isinstance(value, bool):
                if math.isfinite(value):
                    values.append((row_idx, value))

        if not values:
            continue

        min_value = min(value for _, value in values)
        max_value = max(value for _, value in values)

        for row_idx, value in values:
            cell = ws.cell(row_idx, col_idx)

            if is_wastage:
                if value == min_value:
                    cell.fill = green_fill
                    cell.font = Font(color="006100", bold=True)
                if value == max_value:
                    cell.fill = red_fill
                    cell.font = Font(color="9C0006", bold=True)

            elif is_efficiency:
                if value == max_value:
                    cell.fill = green_fill
                    cell.font = Font(color="006100", bold=True)
                if value == min_value:
                    cell.fill = red_fill
                    cell.font = Font(color="9C0006", bold=True)

wb.save(file_path)

print("Performance-based highlighting completed.")
print("Updated workbook:", file_path.resolve())
```
